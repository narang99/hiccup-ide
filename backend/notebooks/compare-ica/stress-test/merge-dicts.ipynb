```python
# need to merge the warmup dicts
from pathlib import Path
import shelve

SHELVE_CACHE_ROOT = Path.cwd() / "pure-case-v3"
SHELVE_CACHE_ROOT.mkdir(parents=True, exist_ok=True)
```

```python
with shelve.open(SHELVE_CACHE_ROOT / "t_inits_warmups") as shelf:
    # run_single_test(DIMS_LIST, ATOMS_RATIO, SAMPLES_RATIO, term3_set, kwarg_key, shelf)
    warmup_metrics = dict(shelf)


with shelve.open(SHELVE_CACHE_ROOT / "t_inits") as shelf:
    # run_single_test(DIMS_LIST, ATOMS_RATIO, SAMPLES_RATIO, term3_set, kwarg_key, shelf)
    metrics = dict(shelf)
```

```python
# all are present
for k in warmup_metrics:
    if k not in metrics:
        print(k)
```

```python
# all are present
f = []
for k in warmup_metrics:
    if k  in metrics:
        # print(warmup_metrics[k], "\n", metrics[k])
        w, m = warmup_metrics[k], metrics[k]
        f.append({"new_mse": w["mse"], "old_mse": m["mse"], "new_sim": w["mean_sim"], "old_sim": m["mean_sim"]})
```
